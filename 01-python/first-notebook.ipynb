## 2. Cells remember each other

Run these two in order.

```python
alerts_per_day = 1098
real_incidents = 99
```

```python
precision = real_incidents / alerts_per_day
print(f"{precision:.1%} of alerts are real")
```

## 3. The trap: run order is not screen order

This is the one thing that catches everybody.

Run the next cell. Then go back and run it **again**.

```python
alerts_per_day = alerts_per_day + 100
alerts_per_day
```

The number went up again. The cell did not "reset" because you ran it twice; it applied twice.

A notebook holds state that survives between runs, so a notebook that looks correct on screen can be in a state that its own code would never produce from a clean start.

**The habit that saves you:** when a result looks strange, use *Runtime, Restart and run all* (Colab) or *Kernel, Restart & Run All* (Jupyter). If it still looks strange, it is a real problem. If it does not, you had stale state.

## 4. Where is this actually running?

```python
import platform, sys

print("Python :", sys.version.split()[0])
print("System :", platform.system(), platform.release())
print("Machine:", platform.machine())
```

That is not your laptop. It is a temporary machine somewhere else, handed to you for the session and thrown away afterwards.

Two consequences worth remembering:

- Files you create disappear when the session ends unless you download them.
- Anything you paste into a notebook goes to that machine. Do not paste credentials, customer data, or anything you would not send to a third party.
