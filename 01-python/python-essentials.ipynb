## 2. A name is a label, not a box

The Python language reference opens its section on naming with four words:
*names refer to objects*.

`id()` gives the identity of the object a name is pointing at. Two names with the
same id are pointing at the same object.

````python
x = 10
y = x

print("x =", x, " y =", y)
print("same object?", id(x) == id(y))
````

`y = x` did not copy anything. It put a second label on the same object.

````python
x = x + 1

print("x =", x, " y =", y)
print("same object?", id(x) == id(y))
````

`x + 1` had no way to change 10, so it built 11 and moved the label. `y` never
moved.

This is why numbers feel like they copy. They do not: they are simply impossible
to change in place, so you never notice the difference.

## 3. Where it stops being harmless

A list *can* be changed in place. Watch what that does to a second label.

````python
scores = [90, 80]
backup = scores

print("one object, two names?", scores is backup)

scores.append(70)

print("scores:", scores)
print("backup:", backup)
````

Nothing assigned to `backup`, and `backup` changed anyway. It was never a backup.

Note the two different questions you can ask:

- `is` asks whether two names point at the same object
- `==` asks whether two objects look the same

````python
a = [1, 2]
b = [1, 2]

print("a == b :", a == b)   # same contents
print("a is b :", a is b)   # different objects
````

## 4. Making a real copy

````python
scores = [90, 80]
backup = list(scores)      # or scores.copy()

scores.append(70)

print("scores:", scores)
print("backup:", backup)
print("same object?", scores is backup)
````

That copies one layer. If the thing holds other things, the inner ones are still
shared.

````python
import copy

grid    = [[1, 2], [3, 4]]
shallow = list(grid)
deep    = copy.deepcopy(grid)

grid[0].append(99)

print("grid:   ", grid)
print("shallow:", shallow)
print("deep:   ", deep)
````

`shallow` is a new outer list holding the *same* inner lists, so the change showed
through. `deepcopy` copied all the way down.

In pandas the same rule applies and the same fix works: `df.copy()`.
